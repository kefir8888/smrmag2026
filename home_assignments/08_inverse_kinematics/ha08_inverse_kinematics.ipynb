## Домашнее задание 8: обратная кинематика

### Задача 0

Поставьте себе симулятор Webots. Посмотрите, какие роботы там есть, как загружать т.н. "мир" (фактически среду и робота), как запускать демо код. Оба задания с роботами нужно будет выполнить именно в этом симуляторе.

### Задача 1

Выведите обратную кинематику для манипулятора с 3 степенями свободы. Это практически та же задача, что была на паре, только теперь плоскость, в которой манипулятор действует, может поворачиваться вместе с ним. 3 степени свободы - это угол вращения вокруг вертикальной оси (угол рыскания/yaw), регулируемый сервомотором в основании манипулятора, и два угла наклона первого сегмента и второго (после "локтя") соответственно.

"Выведите обратную кинематику" - это значит "найдите функцию, переводящую целевые координаты конечного актуатора (конца роборуки) в углы, которые нужно выставить на сервомоторах для достижения этих координат".

Загрузите модель робота Ned. Посмотрите на него, поймите, где он крутится. Попробуйте поуправлять его углами. Разберитесь с демо скриптом. Найдите в интернете документацию к этому роботу, найдите в ней высоту основания, длины сегментов, в общем исчерпывающее описание кинематики робота. Мы пока будем работать с ним как с манипулятором с 3 степенями свободы, но у него их на самом деле 6.

Критерий успешного выполнения задачи - робот может достать до доминошки в сцене в webots.

### Задача 2

Реализуйте численную обратную кинематику для плоского манипулятора (да, прям совсем руками). Нужно будет сделать перебор по решетке (grid search) для обоих углов и выбрать тот вариант, который дает минимальную ошибку среди всех. Ошибка вычисляется как евклидово расстояние между выходом прямой кинематики и целевой точкой. Протестируйте эту функцию, попробуйте нарисовать круг на плоскости с ее помощью.

### Задача 3

Реализуйте численную обратную кинематику для шестиосевого манипулятора на основе _scipy.optimize_. На вход должны поступать целевая позиция и ориентация (в любом удобном формате), на выход 6 углов. Протестируйте ее на Ned.

### Задача 4

Решите аналитически задачу обратной кинематики для платформы Стюарта из Webots. Разберитесь с тем, как управлять этим роботом в симуляции. Вход - координаты центра верхней платформы и ориентация платформы, выход - 6 длин.

### Задача 5

Реализуйте стабилизацию шара на платформе Стюарта с помощью PD-контроллера. Мир в Webots в формате .wbt лежит в репозитории. Нужно будет написать контроллер, который считывает из симулятора положение шара и наклоняет платформу в нужную сторону. Рекомендуемая последовательность шагов:
1) выведите (упрощенную) обратную кинематику для двух углов наклона платформы Стюарта (крен и тангаж, они же roll и pitch) или воспользуйтесь полной из задачи 4
2) запрогайте ее, убедитесь, что платформа дайствительно наклоняется так, как надо
3) найдите разумные значения коэффициентов обратной связи для определения угла наклона в зависимости от отклонения шара
4) сделайте движение шара по квадрату

Движение платформы должно быть достаточно медленным, чтобы не подбрасывать шар. Управление по двум осям можно в приближении считать не связанным (хотя на самом деле они конечно связаны из-за вращательной инерции шара). Можно начать думать про контроллер для системы с одной осью: шарик лежит в желобке, и вы наклоняете его на заданный угол, пытаясь привести в центр. А потом можно сделать то же самое по второй оси.
