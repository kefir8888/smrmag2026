```python
import matplotlib.pyplot as plt

x  = 3
xd = 1
m  = 5
dt = 0.1

iter_num = 1000

xhist, xdhist, uhist, thist = [], [], [], []

Kp, Kd = 1, 1

for i in range(iter_num):
    u = - Kp * x - Kd * xd

    xdd = u / m

    xd += xdd * dt
    x  += xd  * dt

    xhist.append(x)
    xdhist.append(xd)
    uhist.append(u)
    thist.append(i * dt)

plt.plot(thist, xhist)
plt.show()
plt.plot(thist, xdhist)
plt.show()
plt.plot(thist, uhist)
plt.show()
plt.plot(xhist, xdhist)
plt.show()
```
